```python
import pandas as pd
```

```python
movies = pd.read_csv("data/movies.csv")
links = pd.read_csv("data/links.csv")
```

```python
movies_and_links = movies.merge(links, on="movieId")
movies10 = movies_and_links[movies_and_links["tmdbId"].notna()].head(10)
movies10
```

**2.**

```python
import requests
import os
from dotenv import load_dotenv

load_dotenv()

API_KEY = os.getenv("TMDB_API_KEY")
```

```python
def fetch_movie_details(tmdb_id):

    url = f"https://api.themoviedb.org/3/movie/{tmdb_id}"

    params = {
        "api_key": API_KEY
    }

    response = requests.get(url, params=params)

    data = response.json()

    overview = data.get("overview", "")
    homepage = data.get("homepage", "")

    if overview is None:
        overview = ""
    if homepage is None:
        homepage = ""

    return {
        "overview": overview,
        "homepage": homepage
    }
```

```python
fetch_movie_details(231)
```

**3.**

```python
overviews = []
homepages = []

for i in range(len(movies10)):
    tmdb_id = movies10.iloc[i]["tmdbId"]
    
    details = fetch_movie_details(tmdb_id)
    
    overviews.append(details["overview"])
    homepages.append(details["homepage"])

movies10["overview"] = overviews
movies10["homepage"] = homepages
```
